```python
# 1. BRAND STANDARDIZATION

def normalize_brand(value):

    # Handle missing values
    if pd.isna(value):
        return pd.NA

    # Convert value to string and remove leading/trailing spaces
    value = str(value).strip()

    # Remove square brackets and quotation marks
    value = value.strip("[]")
    value = value.replace("'", "").replace('"', "")

    # Normalize multiple spaces
    value = re.sub(r"\s+", " ", value).strip()

    # Convert brand name to lowercase for mapping
    key = value.lower()

    # Define standardized brand names
    brand_mapping = {
        "hp": "HP",
        "asus": "ASUS",
        "acer": "Acer",
        "dell": "Dell",
        "msi": "MSI",
        "lenovo": "Lenovo",
        "apple": "Apple",
        "gigabyte": "Gigabyte",
        "singpc": "SingPC",
        "thunderobot": "Thunderobot"
    }

    # Return standardized brand name
    return brand_mapping.get(key, value)


# Apply brand standardization
df["Brand"] = df["Brand"].apply(normalize_brand)

# 2. HANDLE MISSING BRAND VALUES

def infer_brand(product_name):

    # Handle missing product names
    if pd.isna(product_name):
        return pd.NA

    product_name = str(product_name).lower()

    # Define brand identification patterns
    patterns = {
        "Apple": r"\bmacbook\b",
        "HP": r"\bhp\b",
        "ASUS": r"\basus\b",
        "Dell": r"\bdell\b",
        "Acer": r"\bacer\b",
        "MSI": r"\bmsi\b",
        "Lenovo": r"\blenovo\b",
        "Gigabyte": r"\bgigabyte\b",
        "SingPC": r"\bsingpc\b",
        "Thunderobot": r"\bthunderobot\b"
    }

    # Identify brand based on product name
    for brand, pattern in patterns.items():
        if re.search(pattern, product_name):
            return brand

    # Return missing value if no brand is identified
    return pd.NA


# Identify rows with missing brand values
missing_brand_mask = df["Brand"].isna()

# Infer missing brands from product names
df.loc[missing_brand_mask, "Brand"] = (
    df.loc[missing_brand_mask, "Product Name"]
    .apply(infer_brand)
)


# 3. VALIDATE BRAND STANDARDIZATION


# Display brand distribution
print("Brand Distribution:")
print(df["Brand"].value_counts(dropna=False))

# Count unique brands
print("\nNumber of Unique Brands:")
print(df["Brand"].nunique())

# Check remaining missing values
print("\nMissing Brand Values:")
print(df["Brand"].isna().sum())
```

```python
def detect_cpu_vendor(cpu):

    if pd.isna(cpu):
        return "Missing"

    cpu = cpu.lower()

    if "intel" in cpu or "core i" in cpu:
        return "Intel"

    elif "amd" in cpu or "ryzen" in cpu:
        return "AMD"

    elif "apple" in cpu or re.search(r"\bM[1-9]\b", cpu, re.I):
        return "Apple"

    elif "snapdragon" in cpu or "qualcomm" in cpu:
        return "Qualcomm"

    elif "nvidia" in cpu:
        return "NVIDIA"

    return "Unknown"


df["CPU Vendor"] = df["CPU Clean"].apply(detect_cpu_vendor)

display(df["CPU Vendor"].value_counts().to_frame("Count"))
```

```python

def classify_cpu_family(cpu):

    if pd.isna(cpu):
        return "Unknown"

    cpu = str(cpu).lower().strip()

    patterns = [
        # Intel Core Ultra X Series
        (r"core\s+ultra\s*x([579])", "Intel Core Ultra X"),

        # Intel Core Ultra Series
        (r"core\s+ultra\s*([3579])", "Intel Core Ultra"),

        # Intel Core i Series
        (r"core\s*i([3579])", "Intel Core i"),

        # Intel Core Series (Core 3, 5, 7)
        (r"(?:intel\s+)?core\s+([3579])\b", "Intel Core"),

        # AMD Ryzen AI Max
        (r"ryzen\s+ai\s*max\s*\+", "AMD Ryzen AI Max+"),
        (r"ryzen\s+ai\s*max\b", "AMD Ryzen AI Max"),

        # AMD Ryzen AI
        (r"ryzen\s+ai\s*([3579])", "AMD Ryzen AI"),

        # AMD Ryzen
        (r"ryzen\s+([3579])", "AMD Ryzen"),
    ]

    for pattern, prefix in patterns:

        match = re.search(pattern, cpu)

        if match:
            if "max" in pattern:
                return prefix

            return f"{prefix} {match.group(1)}"

    # Apple M Series
    match = re.search(
        r"\bm([1-9])\s*(pro|max|ultra)?\b",
        cpu
    )

    if match:
        family = f"Apple M{match.group(1)}"

        if match.group(2):
            family += f" {match.group(2).title()}"

        return family

    # Apple A Series
    if re.search(r"\ba\d+\s*(pro|max)?\b", cpu):
        return "Apple A Series"

    # Qualcomm
    if "snapdragon" in cpu or "qualcomm" in cpu:
        return "Qualcomm Snapdragon"

    # NVIDIA
    if "nvidia" in cpu:
        return "NVIDIA"

    return "Other"


df["CPU Family"] = df["CPU Clean"].apply(classify_cpu_family)

display(
    df["CPU Family"]
    .value_counts()
    .to_frame("Count")
)
```

```python
def clean_gpu(gpu):
    if pd.isna(gpu):
        return np.nan

    gpu = str(gpu).strip()

    # 1. Normalize special characters and whitespace
    gpu = re.sub(r"[®™]", "", gpu)
    gpu = re.sub(r"\s+", " ", gpu).strip()

    # 2. Integrated GPU with a specified number of cores
    match = re.search(
        r"Card tích hợp\s*-\s*(?:GPU\s*)?(\d+)\s*(?:nhân|lõi)(?:\s*GPU)?",
        gpu,
        flags=re.IGNORECASE
    )

    if match:
        return f"Integrated GPU - {match.group(1)} cores"

    # 3. Intel Arc integrated graphics
    match = re.search(
        r"Intel Arc\s*(?:Graphics\s*)?\d{3}[A-Z]?",
        gpu,
        flags=re.IGNORECASE
    )

    if match:
        return match.group(0).strip()

    # 4. Intel Graphics integrated GPU
    if re.search(r"Intel Graphics", gpu, flags=re.IGNORECASE):
        return "Intel Graphics"

    # 5. Generic onboard graphics
    if re.fullmatch(
        r"Card onboard\s*\(tích hợp\)",
        gpu,
        flags=re.IGNORECASE
    ):
        return "Integrated Graphics"

    # 6. Generic integrated graphics
    if re.fullmatch(r"Card tích hợp", gpu, flags=re.IGNORECASE):
        return "Integrated Graphics"

    return gpu


df["GPU Clean"] = df["GPU"].apply(clean_gpu)
```

```python
def get_gpu_vendor(row):

    gpu = str(row["GPU Clean"]).lower()
    product = str(row["Product Name"]).lower()
    brand = str(row["Brand"]).lower()

    if "intel" in gpu:
        return "Intel"

    elif "amd" in gpu or "radeon" in gpu:
        return "AMD"

    elif "nvidia" in gpu or "geforce" in gpu:
        return "NVIDIA"

    elif "qualcomm" in gpu or "adreno" in gpu:
        return "Qualcomm"

    elif (
        "integrated gpu" in gpu
        and ("macbook" in product or "apple" in brand)
    ):
        return "Apple"

    else:
        return "Unknown"


df["GPU Vendor"] = df.apply(get_gpu_vendor, axis=1)
```

```python
def get_gpu_family(row):

    gpu = row["GPU Clean"]
    vendor = row["GPU Vendor"]

    if pd.isna(gpu):
        return "Unknown"

    gpu = str(gpu).lower().strip()

    # Apple
    if vendor == "Apple":
        return "Apple GPU"

    # NVIDIA
    elif "geforce rtx" in gpu:
        return "NVIDIA GeForce RTX"

    elif "geforce mx" in gpu:
        return "NVIDIA GeForce MX"

    elif "blackwell" in gpu:
        return "NVIDIA Blackwell"

    # AMD
    elif "radeon" in gpu:
        return "AMD Radeon"

    # Intel
    elif "iris xe" in gpu:
        return "Intel Iris Xe"

    elif "uhd" in gpu:
        return "Intel UHD"

    elif "intel arc" in gpu:
        return "Intel Arc"

    elif "intel graphics" in gpu:
        return "Intel Graphics"

    # Qualcomm
    elif "adreno" in gpu:
        return "Qualcomm Adreno"

    # GPU tích hợp không xác định hãng
    elif "integrated gpu" in gpu:
        return "Integrated GPU"

    elif "integrated graphics" in gpu:
        return "Integrated Graphics"

    else:
        return "Unknown"


df["GPU Family"] = df.apply(get_gpu_family, axis=1)
```

```python
def get_gpu_type(row):

    gpu = row["GPU Clean"]

    if pd.isna(gpu):
        return "Unknown"

    gpu = str(gpu).strip().lower()

    if gpu in ["", "không", "nan"]:
        return "Unknown"

    # 1. Dedicated GPU
    if (
        "card rời" in gpu
        or "geforce rtx" in gpu
        or "geforce mx" in gpu
        or re.search(r"\bradeon rx\b", gpu)
        or ("nvidia" in gpu and "blackwell" in gpu)
    ):
        return "Dedicated"

    # 2. Apple integrated GPU
    if row["GPU Vendor"] == "Apple":
        return "Integrated"

    # 3. Intel integrated GPU
    if any(x in gpu for x in [
        "intel graphics",
        "intel uhd",
        "iris xe",
        "intel arc 140v",
        "intel arc 130v",
        "intel arc 140t",
        "integrated"
    ]):
        return "Integrated"

    # General Intel Arc Graphics
    if "intel arc graphics" in gpu:
        return "Integrated"

    # 4. AMD integrated GPU
    if "radeon" in gpu and (
        "integrated" in gpu
        or "vega" in gpu
        or re.search(r"\bradeon\s+\d{3,4}m\b", gpu)
        or "radeon graphics" in gpu
        or "radeon 8060s" in gpu
    ):
        return "Integrated"

    # 5. Qualcomm Adreno GPU
    if "adreno" in gpu:
        return "Integrated"

    # 6. Onboard or integrated graphics
    if "onboard" in gpu or "card tích hợp" in gpu:
        return "Integrated"

    # 7. Unknown cases that have been manually reviewed
    return "Integrated"


df["GPU Type Clean"] = df.apply(get_gpu_type, axis=1)
```
